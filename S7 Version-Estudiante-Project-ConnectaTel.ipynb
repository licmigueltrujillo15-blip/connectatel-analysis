## 🧩Paso 4: Summary statistics de uso por usuario


### 4.1 Agrupación por comportamiento de uso

🎯**Objetivo**: Resumir las variables clave de la tabla `usage` **por usuario**, creando métricas que representen su comportamiento real de uso histórico. 

**Instrucciones:**: 
1. Construye una tabla agregada de `usage` por `user_id` que incluya:
- número total de mensajes  
- número total de llamadas  
- total de minutos de llamadas

2. Renombra las columnas para que tengan nombres claros:  
- `cant_mensajes`  
- `cant_llamadas`  
- `cant_minutos_llamada`
3. Combina esta tabla con `users`.

```python
# Columnas auxiliares
usage["is_text"] = (usage["type"] == "text").astype(int) #conocer el total de mensajes
usage["is_call"] = (usage["type"] == "call").astype(int) #conocer el total de llamadas


# Agrupar información por usuario
usage_agg = usage.groupby('user_id').agg({
    'is_text': 'sum',
    'is_call': 'sum',
    'duration': 'sum'
}).reset_index()

# observar resultado
usage_agg.head(3)
```

```python
# Renombrar columnas
usage_agg = usage_agg.rename(columns={
    'is_text': 'cant_mensajes',
    'is_call': 'cant_llamadas',
    'duration': 'cant_minutos_llamada'
})

# observar resultado
usage_agg.head(3)
```

```python
# Combinar la tabla agregada con el dataset de usuarios
user_profile = pd.merge(users, usage_agg, on='user_id', how='left')

# Rellenar nulos con 0 (por si algún usuario no tiene uso)
user_profile[['cant_mensajes', 'cant_llamadas', 'cant_minutos_llamada']] = user_profile[['cant_mensajes', 'cant_llamadas', 'cant_minutos_llamada']].fillna(0)

# observar resultado
user_profile.head(5)
```

### 5.2 Identificación de Outliers

🎯 **Objetivo:**  
Detectar valores extremos en las variables clave de **uso** y **clientes** que podrían afectar el análisis, y decidir si requieren limpieza o revisión adicional.

**Instrucciones:**  
- Usa **boxplots** para identificar visualmente outliers en las siguientes columnas:  
  - `age` 
  - `cant_mensajes`
  - `cant_llamadas`
  - `total_minutos_llamada`  
- Crea un **for** para generar los 4 boxplots automáticamente.
<br>

- Después de crear los gráfico, responde si **existen o no outliers** en las variables.  
- Si hay outliers, crea otro bucle para calcular los límites de esas columnas usando el **método IQR** y decide qué hacer con ellos.
  - Si solamente hay outliers de un solo lado, no es necesario calcular ambos límites.

**Hint:**
- Dentro del bucle, usa `plt.title(f'Boxplot: {col}')` para que el título cambie acorde a la columna.

```python
# Visualizando usando BoxPlot
columnas_numericas = ['age', 'cant_mensajes', 'cant_llamadas', 'cant_minutos_llamada']

plt.figure(figsize=(15, 8)) # Hacemos la figura grande para que quepan los 4

for i, col in enumerate(columnas_numericas, 1):
    plt.subplot(2, 2, i) # Crea una cuadrícula de 2x2
    sns.boxplot(y=user_profile[col], color='skyblue')
    plt.title(f'Boxplot: {col}') # Título dinámico como pide el hint
    plt.ylabel('')

plt.tight_layout()
plt.show()
```

💡Insights: 
- Age: No presenta outliers (no hay puntos fuera de los bigotes). Las edades están muy bien distribuidas entre los 18 y 80 años.

- cant_mensajes: Sí presenta outliers en el lado derecho. Hay un grupo de usuarios atípicos que envían entre 12 y 17 mensajes (muy por encima del promedio de 5).

- cant_llamadas: Sí presenta outliers en el lado derecho. Hay usuarios que realizan entre 11 y 15 llamadas, superando el límite superior del bigote (que está en 10).

- cant_minutos_llamada: Sí presenta múltiples outliers en el lado derecho. Este es el caso más extremo: hay un grupo grande de heavy users que consumen entre 70 y 155 minutos, alejados del comportamiento típico (que ronda los 20-30 minutos).

```python
# Calcular límites con el método IQR
columnas_limites = ['cant_mensajes', 'cant_llamadas', 'cant_minutos_llamada']

for col in columnas_limites:
    Q1 = user_profile[col].quantile(0.25)
    Q3 = user_profile[col].quantile(0.75)
    IQR = Q3 - Q1
    
    limite_superior = Q3 + 1.5 * IQR
    
    # Contar cuántos outliers hay
    outliers = user_profile[user_profile[col] > limite_superior]
    
    print(f"--- {col.upper()} ---")
    print(f"Límite superior: {limite_superior:.2f}")
    print(f"Cantidad de outliers: {len(outliers)} usuarios\n")
```

```python
# Revisa los limites superiores y el max, para tomar la decisión de mantener los outliers o no
user_profile[columnas_limites].describe()
```

💡Insights: 
- cant_mensajes: MANTENER los outliers. El límite superior es 11.5 mensajes, y hay 46 usuarios que lo superan (llegando hasta 17). No son errores, son "heavy users" que usan intensivamente el servicio de mensajería. Eliminarlos sesgaría el análisis y perderíamos información sobre los clientes más activos. Los 46 usuarios que lo superan representan solo el 1.15%. Su comportamiento es real y no un error de captura.

- cant_llamadas: MANTENER los outliers. El límite superior es 10.5 llamadas, con 30 usuarios por encima (llegando a 15). Al igual que con los mensajes, representan un comportamiento real de usuarios que dependen altamente de las llamadas de voz. Son candidatos ideales para ofrecerles planes con más beneficios. Los 30 usuarios atípicos representan el 0.75% de la base.

- cant_minutos_llamada: MANTENER los outliers. Aquí es donde hay más atípicos: 109 usuarios superan el límite de 61.87 minutos, con un máximo impresionante de 155.69 minutos. Estos clientes son el segmento de mayor valor (probablemente usuarios Premium). Borrarlos no solo sería un error estadístico, sino una ceguera de negocio que nos impediría diseñar estrategias de retención para los clientes que más consumen (y que más pagan). Los 109 usuarios atípicos representan el 2.72% de la base.

**Instrucciones:**

- Usa comparaciones lógicas (<, >) para evaluar las condiciones de llamadas y mensajes y asigna:
  - `'Bajo uso'` cuando llamadas < 5 y mensajes < 5
  - `'Uso medio'` cuando llamadas < 10 y mensajes < 10
  - `'Alto uso'` para el resto de casos

```python
import numpy as np

# Crear condiciones y valores para la nueva columna
condiciones = [
    (user_profile['cant_llamadas'] < 5) & (user_profile['cant_mensajes'] < 5),  # Bajo uso
    (user_profile['cant_llamadas'] < 10) & (user_profile['cant_mensajes'] < 10) # Uso medio
]
valores = ['Bajo uso', 'Uso medio']

# Crear columna grupo_uso
# np.select evalúa las condiciones en orden. Si no cumple ninguna, usa el 'default'
user_profile['grupo_uso'] = np.select(condiciones, valores, default='Alto uso')
```
